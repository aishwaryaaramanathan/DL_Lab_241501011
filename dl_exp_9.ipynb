```python
import tensorflow as tf
import numpy as np
```

```python
# Load Shakespeare dataset
path = tf.keras.utils.get_file(
    'shakespeare.txt',
    'https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt'
)
text = open(path, encoding='utf-8').read().lower()
```

```python
# Encode characters
chars = sorted(set(text))
c2i = {c: i for i, c in enumerate(chars)}
i2c = {i: c for i, c in enumerate(chars)}
```

```python
# Create sequences
seq_len = 40
X, y = [], []

for i in range(len(text) - seq_len):
    X.append([c2i[c] for c in text[i:i+seq_len]])
    y.append(c2i[text[i+seq_len]])

X = np.array(X)
y = np.array(y)
```

```python
# Build and train LSTM
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(seq_len,)),
    tf.keras.layers.Embedding(len(chars), 64),
    tf.keras.layers.LSTM(128),
    tf.keras.layers.Dense(len(chars), activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy')

model.fit(X, y, epochs=5, batch_size=128)
```

```python
# Generate text
def generate(seed, length=200):
    seed = seed.lower()
    seq = [c2i.get(c, c2i[' ']) for c in seed]

    result = seed
    for _ in range(length):
        inp = np.array(seq[-seq_len:]).reshape(1, -1)
        pred = model.predict(inp, verbose=0)[0]
        idx = np.argmax(pred)
        seq.append(idx)
        result += i2c[idx]

    return result

print("Generated Text:\n")
print(generate("shall i compare thee to a summer's day?\n"))
```
