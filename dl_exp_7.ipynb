```python
import torch
from transformers import VisionEncoderDecoderModel, ViTImageProcessor, AutoTokenizer
from PIL import Image
```

```python
# Load pretrained model
name = "nlpconnect/vit-gpt2-image-captioning"
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = VisionEncoderDecoderModel.from_pretrained(name).to(device)
processor = ViTImageProcessor.from_pretrained(name)
tokenizer = AutoTokenizer.from_pretrained(name)
model.eval()
```

```python
# Generate image caption
def generate_caption(path):
    image = Image.open(path).convert("RGB")
    pixels = processor(images=image, return_tensors="pt").pixel_values.to(device)

    with torch.no_grad():
        output = model.generate(pixels, max_length=30, num_beams=4)

    return tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
# Test with an image
path = "work.jpeg"  # Replace with your image path
print("Generated Caption:", generate_caption(path))
```
